# Objetivos y escenario de la práctica

## Objetivo general

Aplicar las fases de comprensión del negocio y comprensión de los datos de CRISP-DM, y desarrollar la preparación de datos hasta la normalización Min-Max y la estandarización Z-score, mediante el análisis reproducible de más de 100 000 registros de hospitalizaciones de pacientes con diabetes en Python.

## Antecedentes

Se utilizará el conjunto de datos *Diabetes 130-US Hospitals for Years 1999–2008*, publicado por UCI Machine Learning Repository.

El conjunto contiene 101 766 hospitalizaciones registradas en 130 hospitales y redes de atención de Estados Unidos.

Cada fila representa una hospitalización, no necesariamente un paciente diferente.

## Planteamiento del problema

Un equipo hospitalario desea explorar cómo se distribuyen las características de las hospitalizaciones de pacientes con diabetes y cómo se relacionan descriptivamente con la readmisión.

Antes de desarrollar modelos, es necesario conocer la calidad de los datos y preparar versiones comparables de las variables cuantitativas.

## Alcance

- Fase 1 de CRISP-DM: comprensión del negocio.
- Fase 2 de CRISP-DM: comprensión de los datos.
- Fase 3 de CRISP-DM: selección, limpieza y escalamiento.

No se entrenarán ni evaluarán modelos predictivos. El análisis es académico y no permite emitir diagnósticos ni establecer relaciones causales.
