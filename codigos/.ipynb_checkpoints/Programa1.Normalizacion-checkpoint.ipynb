**Normalización:** Es una técnica de preparación de datos que transforma los valores numéricos a una escala determinada, generalmente entre **0 y 1**. Esto permite trabajar con variables que originalmente tienen diferentes escalas y evita que unas tengan más influencia que otras solamente por su tamaño numérico.

### Fórmula:

$$
X_{ij}^{N} = \left\{ \frac{X_{ij} - \min(X_i)}{\max(X_i) - \min(X_i)}, \quad \forall j \in \{1,\ldots,n\} \right\}
$$

Donde:

- **X:** valor original.
- **X mínimo:** valor mínimo de la columna.
- **X máximo:** valor máximo de la columna.
- **X normalizado:** valor transformado entre 0 y 1.


```python
from sklearn import preprocessing
import numpy as np
```

**sklearn.preprocessing:** Se utiliza para la preparación y transformación de datos antes de entrenar un modelo de Machine Learning.

**numpy:** Se utiliza para trabajar con vectores, matrices y arreglos numéricos, además de realizar operaciones matemáticas de manera eficiente.

```python
# Carga de datos
X_train = np.array([
    [1., -4., 5.],
    [5.,  0., 2.],
    [0.,  6., -3.]
])

print(X_train)
```

```python
# Limpieza de datos: normalización
normalizer = preprocessing.MinMaxScaler()
X_train_norm = normalizer.fit_transform(X_train)
```

**MinMaxScaler():** Es una herramienta de Scikit-learn que sirve para transformar los datos a una escala entre 0 y 1.

**fit():** Calcula y guarda los valores mínimos y máximos de cada columna del conjunto de entrenamiento.

**transform():** Usa esos valores mínimos y máximos para transformar los datos.

**fit_transform():**
Hace las dos cosas en un solo paso:
1. aprende los mínimos y máximos;
2. transforma los datos.
El resultado se guarda en **`X_train_norm`**.

```python
# Salida de resultados
print(X_train_norm)
```

```python
# Ejemplo para ver que los resultados son iguales
# si aplicamos los métodos "fit" y "transform" por separado

X_train_2 = np.copy(X_train)

# Crear un objeto ajustado con fit
f = normalizer.fit(X_train)

# Aplicar transform sobre la copia
X_train_norm_2 = f.transform(X_train_2)

# Comprobar si son iguales
print(np.array_equiv(X_train_norm, X_train_norm_2))
```

```python
# Normalizar un nuevo conjunto de datos de test
X_test = np.array([
    [1., -3.,  2.],
    [3.,  0.,  0.],
    [0.,  4., -1.]
])

# Aplicar únicamente transform
X_test_norm = f.transform(X_test)
```

Aquí se crea una nueva matriz llamada **X_test**, que representa datos de prueba.
Para normalizarlos se utiliza solamente:

### `transform()`

Esto se hace porque los valores mínimos y máximos ya fueron aprendidos con **X_train**.
No se debe volver a usar **fit()** con los datos de prueba, porque eso haría que el modelo utilice una escala diferente.
Así, los datos de prueba quedan transformados con la **misma regla** que los datos de entrenamiento.

```python
print(X_test_norm)
```
