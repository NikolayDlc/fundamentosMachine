La **estandarización** es una técnica de preparación de datos que transforma los valores numéricos tomando en cuenta su media y su desviación estándar. Su objetivo es que cada característica tenga una media de 0 y una desviación estándar de 1, siempre que la característica tenga variación. A diferencia de la normalización Min-Max, la estandarización no limita los valores al intervalo de 0 a 1.

### Fórmula de estandarización Z-score

$$
Z = \frac{X-\mu}{\sigma}
$$

Donde:

- **Z:** valor estandarizado.
- **X:** valor original.
- **μ (mu):** media de los datos.
- **σ (sigma):** desviación estándar de los datos.

```python
from sklearn import preprocessing
import numpy as np
```

**Scikit-learn (sklearn):** Es una librería de Python que proporciona herramientas para trabajar con Machine Learning.

**preprocessing:** Es un módulo de Scikit-learn utilizado para preparar y transformar los datos antes de entrenar modelos.

**NumPy (np):** Es una librería que permite trabajar con arreglos, vectores y matrices numéricas. 


```python

# Carga de datos

X_train = np.array([
    [1., -1.,  2.],
    [2.,  0.,  0.],
    [0.,  1., -1.]
])
```

```python

# Limpieza de datos: estandarización

standardizer = preprocessing.StandardScaler()

# Estandarizamos los datos de entrenamiento
X_train_std = standardizer.fit_transform(X_train)

print(X_train_std)
```

**StandardScaler():** Es una herramienta de Scikit-learn que permite estandarizar las características numéricas utilizando su media y desviación estándar.

**standardizer:** Es el objeto que utilizaremos para realizar la estandarización.

**fit():** Calcula y almacena la media y la desviación estándar de cada característica de los datos de entrenamiento.

**transform():** Transforma los datos utilizando la media y la desviación estándar calculadas anteriormente.

**fit_transform():** Combina las dos operaciones anteriores: primero calcula los parámetros y después transforma los datos.

**X_train_std:** Es la matriz que contiene los datos estandarizados.

Los valores resultantes pueden ser negativos, positivos o iguales a cero.


```python

# Crear una segunda matriz con los mismos datos

X_train_2 = np.array([
    [1., -1.,  2.],
    [2.,  0.,  0.],
    [0.,  1., -1.]
])
```

```python

# Aplicamos los métodos "fit" y "transform"
# por separado

f = preprocessing.StandardScaler()

f = f.fit(X_train)

X_train_2_std = f.transform(X_train_2)

print(X_train_2_std)
```


**f = preprocessing.StandardScaler():** Crea un nuevo objeto encargado de estandarizar los datos.

**f.fit(X_train):** Calcula y almacena la media y la desviación estándar de los datos de entrenamiento.

**f.transform(X_train_2):** Utiliza los parámetros obtenidos anteriormente para transformar los valores de la segunda matriz.

**print(X_train_2_std):** Muestra los resultados estandarizados.

```python

# Comprobar si las matrices estandarizadas son iguales

print(np.array_equiv(X_train_std, X_train_2_std))
```

**np.array_equiv():** Es una función de NumPy que permite comprobar si dos arreglos tienen la misma forma y contienen los mismos valores.
